# Python Lists Exercises


### Exercise 1
Create a list called `fruits` containing the following values: `"apple"`, `"banana"`, `"orange"`, and `"grape"`. Print the list.

```python
fruits = ['apple', 'banana', 'orange', 'grape']
print(fruits)
```

### Exercise 2

```python
numbers = [4, 8, 15, 16, 23, 42]
print(numbers[0])
```

### Exercise 3
Given the list `numbers`, print the last element of the list.

```python
numbers = [4, 8, 15, 16, 23, 42]
print(numbers[-1])
```

### Exercise 4

### Exercise 5
Create a list called `cities` with five city names. Print the number of elements in the list.

```python
cities = ['Bogota', 'Medellin', 'Cali', 'Cartagena', 'Barranquilla']
print(len(cities))
```

### Exercise 6
Given the list `animals`, add `"hamster"` to the end of the list and print the updated list.

```python
animals = ['dog', 'cat', 'rabbit']
animals.append('hamster')
print(animals)
```

### Exercise 7

```python
animals = ['dog', 'cat', 'rabbit']
animals.insert(1, 'bird')
print(animals)
```

### Exercise 8

### Exercise 9

### Exercise 10

### Exercise 11
Given the list `numbers`, sort the list in ascending order and print the result.

```python
numbers = [5, 2, 9, 1, 7]
numbers.sort()
print(numbers)
```

### Exercise 12
Given the list `numbers`, sort the list in descending order and print the result.

```python
numbers = [5, 2, 9, 1, 7]
numbers.sort(reverse=True)
print(numbers)
```

### Exercise 13
Given the list `numbers`, create a new list containing each number multiplied by 2.

```python
numbers = [3, 6, 9, 12, 15]
doubled = [n * 2 for n in numbers]
print(doubled)
```

### Exercise 14
Given the list `words`, use a loop to print each word on a separate line.

```python
words = ['python', 'list', 'exercise', 'practice']
for word in words:
    print(word)
```

### Exercise 15
Given the list `grades`, calculate and print the average grade.

```python
grades = [80, 75, 90, 85, 95]
average = sum(grades) / len(grades)
print(average)
```
